```python
import urllib.request as req
import bs4 as bs

def get_all_images_href(url):
    # url = "https://www.ptt.cc/bbs/Beauty/M.1765471034.A.69D.html"
    resp = req.urlopen(url)
    content = resp.read()
    html = bs.BeautifulSoup(content)
    # print(html)
    # no class:
    # 1. find parent
    # 2. custom
    links = html.find_all("a")

    result = []
    allow_sub = ["jpg", "jpeg", "png", "gif"]
    for l in links:
        l_href = l["href"]
        sub = l_href.split(".")[-1]
        if sub.lower() in allow_sub:
            result.append(l_href)
    return result

get_all_images_href("https://www.ptt.cc/bbs/Beauty/M.1765471034.A.69D.html")
```
