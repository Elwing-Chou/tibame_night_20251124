```python
# pip install undetected-chromedriver
# pip install --upgrade setuptools

import os
import time
import urllib.request as req
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

# selenium -> driver(命令-實際操控轉接) -> chrome
# 因為driver就是瀏覽器公司開發的 所以很容易被擋
import undetected_chromedriver as uc

# 如果你是MAC, req.urlopen 遇到SSL Certificate Failed: 就加這兩行
import ssl
ssl._create_default_https_context = ssl._create_unverified_context

# 你要把你所有的平常動作模擬在這裡
if __name__ == '__main__':
    driver = uc.Chrome(version_main=144, use_subprocess=False)
    driver.get("https://www.google.com")
    driver.maximize_window()

    # 跟他說哪個html區塊(BeautifulSoup)
    # bs: find/find_all  tag+class/id
    # selenium: find_element/find_elements  tag/class/id
    e = driver.find_element(By.CLASS_NAME, "gLFyf")
    # click: 滑鼠  send_keys: 鍵盤
    e.send_keys("烏薩奇")
    e.send_keys(Keys.ENTER)
    time.sleep(3)

    es = driver.find_elements(By.CLASS_NAME, "C6AK7c")
    es[2].click()
    time.sleep(3)

    dn = "chiikawa"
    if not os.path.exists(dn):
        os.makedirs(dn)

    es = driver.find_elements(By.CLASS_NAME, "YQ4gaf")
    for img in es:
        # 每種型態可以擁有: 1. 專屬功能(吃飯()) 2. 專屬值(身高)
        # bs萃取: ["href"]/.get_text() or .text
        # selenium: .get_attribute("href")/.text
        img_url = img.get_attribute("src")
        fn = dn + "/" + str(time.time()) + ".jpg"
        req.urlretrieve(img_url, fn)

    time.sleep(3)
```

```python
# cookie: 儲存在瀏覽器, 下次被自動帶入header的值
# 有些往只在第一次和第二次會得到不同回應
# e.g. google/fb登入
# 送一樣的->依樣的回應
# 會不一樣->Headers不一樣
# Cookie: 在第一次得到回應以後, 伺服器會要求你在瀏覽器裡面存一些值(Cookie)
#         下次再拜訪一樣網站, 自動把這些cookie帶入headers裡

# pip install undetected-chromedriver
# pip install --upgrade setuptools

import os
import time
import urllib.request as req
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys

# selenium -> driver(命令-實際操控轉接) -> chrome
# 因為driver就是瀏覽器公司開發的 所以很容易被擋
import undetected_chromedriver as uc

# 如果你是MAC, req.urlopen 遇到SSL Certificate Failed: 就加這兩行
import ssl
ssl._create_default_https_context = ssl._create_unverified_context

# 你要把你所有的平常動作模擬在這裡
if __name__ == '__main__':
    driver = uc.Chrome(version_main=144, use_subprocess=False)
    driver.get("https://www.facebook.com")
    driver.maximize_window()

    cookie_s = '你要去拿自己的'
    for c in cookie_s.split(";"):
        c = c.strip()
        k, v = c.split("=", 1)
        print(k, v)
        driver.add_cookie({"name": k, "value": v})

    driver.refresh()
    time.sleep(3)
```

```python
# python with sql
import sqlalchemy
from sqlalchemy import create_engine
from sqlalchemy import Table, Column, Integer, String, MetaData, ForeignKey
from sqlalchemy import inspect
from sqlalchemy.sql import text


metadata = MetaData()
books = Table('book', metadata,
  Column('id', Integer, primary_key=True),
  Column('title', String),
  Column('primary_author', String),
)

engine = create_engine('sqlite:///bookstore.db')
metadata.create_all(engine)

# insert
with engine.connect() as connection:
    title = "Best Journey"
    primary_author = "Elwing"

    statement = text("""INSERT INTO book(title, primary_author) VALUES(:title, :primary_author)""")

    connection.execute(statement.params(title=title, primary_author=primary_author))
    connection.commit()

    result = connection.execute(text('SELECT * FROM book'))
    for row in result:
        print(row)
```
